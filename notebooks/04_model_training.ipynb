```python
#Train/test split
from sklearn.model_selection import train_test_split

x = data.drop("Churn", axis=1) #Features
y = data["Churn"] #Target

x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42)
```

```python
# Importing classification models to be used in the analysis

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

model = {
    "Logistic Regression": LogisticRegression(max_iter=10000),
  "Decision Tree": DecisionTreeClassifier(max_depth=5, min_samples_leaf=40,class_weight='balanced'),
  "Random Forest": RandomForestClassifier(
    n_estimators=100,
    max_depth=10,
    min_samples_split=10,
    min_samples_leaf=5,
    max_features='sqrt',
    random_state=42,
    class_weight="balanced"
)}
```

```python
# Importing metrics for evaluating classification models
from sklearn.metrics import accuracy_score, classification_report
```

```python
# Training and evaluating each model

for name, model in model.items():
    model.fit(x_train, y_train)
    y_pred = model.predict(x_test)

    print(f"\n🔹 {name}")
    print("Accuracy:",accuracy_score(y_test, y_pred))
    print(classification_report(y_test, y_pred))
    print("-"*60)
```

## Model Training Conclusion

Three models were evaluated to predict customer churn: Logistic Regression, Decision Tree, and Random Forest.

Logistic Regression achieved the best overall performance with an accuracy of approximately 81.6%, showing a good balance between precision and recall.

However, the Decision Tree and Random Forest models showed higher recall for the churn class, meaning they are slightly better at identifying customers who are likely to leave.

This indicates that while Logistic Regression is the most stable overall model, tree-based models may be more useful when the business priority is to catch as many churned customers as possible.
