# Churn Analysis

This section analyzes customer churn patterns.

```python
# Reviewing churn distribution to understand how many customers left versus retained
data["Churn"].value_counts()
```

```python
#Average Monthly Charges by Churn
data.groupby("Churn")["MonthlyCharges"].mean().round(2)
```

```python
#Tenure vs Churn

#Creating tenure groups
data["tenure_group"] = pd.cut(
    data["tenure"],
    bins=[0, 12, 24, 60, 100],
    labels=["0-12", "12-24", "24-60", "60+"])

#Analyze Churn by Tenure
churn_tenure = pd.crosstab(
    data["tenure_group"],
    data["Churn"],
    normalize="index"
) * 100

churn_tenure

```

```python
#Total Charges vs Churn
data.groupby("Churn")["TotalCharges"].mean().round(2)
```

```python
#High-risk customers
#Selecting customers with tenure < 6 months and monthly charges higher than median
high_risk = data[
    (data["tenure"] < 6) &
    (data["MonthlyCharges"] > data["MonthlyCharges"].median())
]

high_risk.head()
```

```python
#Group size
high_risk.shape[0]
```

```python
#Churn rate for high_risk  group
(high_risk["Churn"].value_counts(normalize=True) * 100).round(0)
```

```python
#Comparing with overall churn
(data["Churn"].value_counts(normalize=True) * 100).round(0)
```

Customers with low tenure and high monthly charges show a higher likelihood of churn compared to the overall population.

This suggests that pricing sensitivity is especially important for new customers, and early-stage retention strategies could reduce churn.
